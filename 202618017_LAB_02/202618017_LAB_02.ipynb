```python
import numpy as np

square_matrix = np.array([[4, 7], [2, 6]])
print(f"Square Matrix:\n{square_matrix}")

matrix_transpose = square_matrix.T
print(f"Transpose of Square Matrix:\n{matrix_transpose}")
matrix_determinant = int(np.linalg.det(square_matrix))

print(f"Determinant of Square Matrix: {matrix_determinant}")

if matrix_determinant != 0:
    matrix_inverse = np.linalg.inv(square_matrix)
    print(f"Inverse of Square Matrix:\n{matrix_inverse}")

    # Verify A @ A^-1 using np.allclose()
    identity_matrix = square_matrix @ matrix_inverse

    print(f"A @ A^-1:\n{np.round(identity_matrix, 2)}")
else:
    print("The square matrix is singular and does not have an inverse.")
```

### Missing Values and Fare Outliers

#### Find the missing-value count and percentage for every column.

```python
# Calculate missing value counts
missing_count = data.isnull().sum()

# Calculate missing value percentages
missing_pct = (data.isnull().sum() / len(data)) * 100

# Create a DataFrame to display missing info
missing_df = pd.DataFrame({'Missing Count': missing_count, 'Percentage (%)': missing_pct})
display(missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False))
```

#### Plot missing counts as a bar chart.

```python
# Filter for columns with missing values
missing_data_to_plot = missing_df[missing_df['Missing Count'] > 0]

plt.figure(figsize=(10, 6))
sns.barplot(x=missing_data_to_plot.index, y=missing_data_to_plot['Missing Count'], palette='viridis')
plt.title('Missing Value Counts per Column')
plt.xlabel('Columns')
plt.ylabel('Missing Count')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

#### Fill missing Age values with the mean Age. Show missing counts before and after.

```python
# Create a copy of the dataframe to avoid modifying the original 'data' DataFrame directly for this step
df_mean_fill = data.copy()

# Show missing Age values BEFORE filling
display(f"Missing 'Age' values before mean imputation: {df_mean_fill['Age'].isnull().sum()}")

# Calculate the mean age
mean_age = df_mean_fill['Age'].mean()

# Fill missing Age values with the mean age
df_mean_fill['Age_Mean_Filled'] = df_mean_fill['Age'].fillna(mean_age)

# Show missing Age values AFTER filling
display(f"Missing 'Age' values after mean imputation: {df_mean_fill['Age_Mean_Filled'].isnull().sum()}")

# Display head of df_mean_fill to show the new column
display(df_mean_fill[['Age', 'Age_Mean_Filled']].head())
```

#### Try to fill NaN values by imputing using: mean, median, mode, random value

```python
# Create a fresh copy of the dataframe for various imputation methods
df_impute = data.copy()

# Imputation using Mean (for 'Age')
mean_age = df_impute['Age'].mean()
df_impute['Age_Mean'] = df_impute['Age'].fillna(mean_age)

# Imputation using Median (for 'Age')
median_age = df_impute['Age'].median()
df_impute['Age_Median'] = df_impute['Age'].fillna(median_age)

# Imputation using Mode (for 'Embarked')
# For categorical data, mode is typically used. Handle potential multiple modes by picking the first one.
mode_embarked = df_impute['Embarked'].mode()[0]
df_impute['Embarked_Mode'] = df_impute['Embarked'].fillna(mode_embarked)

# Imputation using Random Value (for 'Age')
# For Age, we'll draw random samples from the non-missing age values.
nan_mask = df_impute['Age'].isnull()
age_choices = df_impute['Age'].dropna().values
num_missing_age = nan_mask.sum()

# Only sample if there are missing values to fill
if num_missing_age > 0:
    df_impute.loc[nan_mask, 'Age_Random'] = np.random.choice(age_choices, size=num_missing_age)
else:
    df_impute['Age_Random'] = df_impute['Age'] # No missing values, so just copy

# Display missing counts after various imputations for the relevant columns
print("\nMissing values after various imputations:")
print(f"Age (Mean): {df_impute['Age_Mean'].isnull().sum()}")
print(f"Age (Median): {df_impute['Age_Median'].isnull().sum()}")
print(f"Embarked (Mode): {df_impute['Embarked_Mode'].isnull().sum()}")
print(f"Age (Random): {df_impute['Age_Random'].isnull().sum()}")

# Display a sample of the imputed columns
display(df_impute[['Age', 'Age_Mean', 'Age_Median', 'Age_Random', 'Embarked', 'Embarked_Mode']].sample(10, random_state=42))
```

```python
# Calculate Q1 (25th percentile) for 'Fare'
Q1 = data['Fare'].quantile(0.25)

# Calculate Q3 (75th percentile) for 'Fare'
Q3 = data['Fare'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Calculate the upper and lower bounds for outlier detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

display(f"Q1 (25th percentile) of Fare: {Q1:.2f}")
display(f"Q3 (75th percentile) of Fare: {Q3:.2f}")
display(f"IQR (Interquartile Range) of Fare: {IQR:.2f}")
display(f"Lower Bound for Outliers: {lower_bound:.2f}")
display(f"Upper Bound for Outliers: {upper_bound:.2f}")

# Count the number of outliers
outliers = data[(data['Fare'] < lower_bound) | (data['Fare'] > upper_bound)]
outlier_count = len(outliers)

display(f"Number of Fare outliers: {outlier_count}")
display(f"Percentage of Fare outliers: {(outlier_count / len(data) * 100):.2f}%")

# Display the first few rows of outliers (optional)
display("Examples of Fare Outliers:", outliers.head())
```

* Age vs. Fare Scatter Plot (Colored by Survival Status): This scatter plot shows the distribution of passengers based on their age and the fare they paid, with different colors indicating whether they survived or not. It helps visualize potential clusters or trends. For instance, you might observe that a higher proportion of survivors tend to be concentrated in regions with higher fares, regardless of age, reinforcing the idea that economic class (often reflected in fare) played a role in survival.
